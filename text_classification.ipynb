```python
categories_mapping = {
    "Fiction": "Fiction",
    "Juvenile Fiction": "Children's Fiction",
    "Biography & Autobiography": "Nonfiction",
    "History": "Nonfiction",
    "Literary Criticism": "Nonfiction",
    "Philosophy": "Nonfiction",
    "Religion": "Nonfiction",
    "Comics & Graphic Novels": "Fiction",
    "Drama": "Fiction",
    "Juveline Nonfiction": "Children's Nonfiction",
    "Science": "Nonfiction",
    "Poetry": "Nonfiction"
}
```

```python
books["simple_categories"] = books["categories"].map(categories_mapping)
```

```python
books[~books["simple_categories"].isna()]
```

```python
from transformers.pipelines import pipeline
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli", device="cuda")

categories = ["Fiction", "Nonfiction"]
```

```python
sequence = books.loc[books["categories"] == "Fiction", "description"].reset_index(drop=True)[0]
```

```python
classifier(sequence, categories)
```

```python
max_index = np.argmax(classifier(sequence, categories)["scores"])
max_label = classifier(sequence, categories)["labels"][max_index]
max_label
```

```python
def generate_predictions(sequence, categories):
    predictions = classifier(sequence, categories)
    max_index = np.argmax(predictions["scores"])
    max_label = predictions["labels"][max_index]
    return max_label
```

```python
from tqdm import tqdm

actual_cats = []
predicted_cats = []

for i in tqdm(range(0, 300)):
    sequence = books.loc[books["simple_categories"] == "Fiction", "description"].reset_index(drop=True)[i]
    predicted_cats += [generate_predictions(sequence, categories)]
    actual_cats += ["Fiction"]
```

```python
for i in tqdm(range(0, 300)):
    sequence = books.loc[books["simple_categories"] == "Nonfiction", "description"].reset_index(drop=True)[i]
    predicted_cats += [generate_predictions(sequence, categories)]
    actual_cats += ["Nonfiction"]
```

```python
predictions_df = pd.DataFrame({"actual_categories": actual_cats, "predicted_categories": predicted_cats})
predictions_df
```

```python
predictions_df["correct_predictions"] = np.where(predictions_df["actual_categories"] == predictions_df["predicted_categories"], 1, 0)
```

```python
# checking predictions accuracy
predictions_df["correct_predictions"].sum()/len(predictions_df)
```

```python
isbns = []
predicted_cats = []

missing_cats = books.loc[books["simple_categories"].isna(), ["isbn13", "description"]].reset_index(drop=True)
```

```python
for i in tqdm(range(0, len(missing_cats))):
    sequence = missing_cats["description"][i]
    predicted_cats += [generate_predictions(sequence, categories)]
    isbns += [missing_cats["isbn13"][i]]
```

```python
missing_predictions_df = pd.DataFrame({"isbn13": isbns, "predicted_categories": predicted_cats})
missing_predictions_df
```

```python
books = pd.merge(books, missing_predictions_df, on="isbn13", how="left")
books["simple_categories"] = np.where(
    books["simple_categories"].isna(), # condition: simple_categories = NaN
    books["predicted_categories"], # True: fill with predicted category
    books["simple_categories"] # False: fill with actual category
)
```

```python
books = books.drop(columns=["predicted_categories"])
books
```
